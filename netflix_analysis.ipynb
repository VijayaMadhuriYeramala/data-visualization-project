```python
import pandas as pd

df = pd.read_csv("netflix_titles.csv")

df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

type_counts = df['type'].value_counts()

sns.barplot(x=type_counts.index, y=type_counts.values)

plt.title("Movies vs TV Shows")
plt.xlabel("Type")
plt.ylabel("Count")

plt.show()
```

```python
# Top Countries (Seaborn)

top_countries = df['country'].value_counts().head(10)

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

sns.barplot(x=top_countries.values, y=top_countries.index)

plt.title("Top 10 Countries Producing Content")
plt.xlabel("Count")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
df['date_added'] = df['date_added'].str.strip()
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df = df.dropna(subset=["date_added"])
df['year'] = df['date_added'].dt.year
```

```python
# Trend over time

year_counts = df['year'].value_counts().sort_index()

plt.figure(figsize=(10,5))

sns.lineplot(x=year_counts.index, y=year_counts.values, marker='o')

plt.title("Content Added Over Time")
plt.xlabel("Year")
plt.ylabel("Count")

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Pie chart (Matplotlib, styled)

type_counts = df['type'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(type_counts, labels=type_counts.index, autopct='%1.0f%%')
plt.title("Movies vs TV Shows")
plt.show()
```

```python
# Better Insights

top_country = df['country'].value_counts().idxmax()
top_year = df['year'].value_counts().idxmax()
content_split = df['type'].value_counts(normalize=True) * 100

print("Key Insights:\n")

print(f"- {content_split.index[0]} dominates with {content_split.iloc[0]:.1f}% of content")
print(f"- Top producing country: {top_country}")
print(f"- Peak content addition year: {top_year}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Product": ["A", "B", "C", "A", "B"],
    "Sales": [450, 650, 400, 400, 350],
    "Region": ["East", "West", "East", "North", "West"]
}

df = pd.DataFrame(data)
df
```

```python
# Product-wise sales (Bar Chart)
product_sales = df.groupby("Product")["Sales"].sum()
product_sales.plot(kind='bar')

plt.title("Product-wise")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

```python

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
sales = [1200, 1500, 1700, 1600, 1800, 2000,
         2200, 2100, 2300, 2400, 2600, 3000]

# Create line chart
plt.figure(figsize=(8,5))
plt.plot(months, sales, marker='o', linewidth=2)
plt.title("Sales Trend Over Time")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

region_sales.plot(kind='pie', autopct='%1.0f%%')

plt.title("Sales Distribution by Region")
plt.ylabel("")
plt.show()
```

```python
product_sales = df.groupby("Product")["Sales"].sum()

ax = product_sales.plot(kind='bar')

product_sales.plot(kind='bar', color=['blue', 'green', 'orange'])

plt.title("Product-wise Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()

for i, v in enumerate(product_sales):
      plt.text(i, v + 20, str(v), ha='center')
plt.show()
```
